````python
import pandas as pd

df = pd.read_csv('customer_shopping_behavior.csv')
df.head()
````

````python
import pandas as pd
import numpy as np

total_revenue = df['Purchase Amount (USD)'].sum()
avg_order_value = df['Purchase Amount (USD)'].mean()
total_transactions = len(df)

category_revenue = df.groupby('Category')['Purchase Amount (USD)'].sum().sort_values(ascending=False)
category_pct = (category_revenue / total_revenue * 100).round(1)

season_revenue = df.groupby('Season')['Purchase Amount (USD)'].sum().sort_values(ascending=False)

gender_revenue = df.groupby('Gender')['Purchase Amount (USD)'].sum()
subscription_avg = df.groupby('Subscription Status')['Purchase Amount (USD)'].mean()

mean_amt = df['Purchase Amount (USD)'].mean()
std_amt = df['Purchase Amount (USD)'].std()
threshold = mean_amt + 2 * std_amt
anomalies = df[df['Purchase Amount (USD)'] > threshold]

print(f"Total Revenue: ${total_revenue:,.2f}")
print(f"Avg Order Value: ${avg_order_value:.2f}")
print(f"Total Transactions: {total_transactions}\n")
print("Revenue by Category:\n", category_revenue, "\n")
print("% of Revenue by Category:\n", category_pct, "\n")
print("Revenue by Season:\n", season_revenue, "\n")
print("Revenue by Gender:\n", gender_revenue, "\n")
print("Avg Order Value by Subscription Status:\n", subscription_avg, "\n")
print(f"Anomaly threshold: ${threshold:.2f}")
print(f"Number of anomalous transactions: {len(anomalies)}")
print(anomalies[['Customer ID', 'Category', 'Purchase Amount (USD)']])
````

````python
import json

# Build a compact summary of your real numbers to feed the AI
stats_summary = f"""
Total Revenue: ${total_revenue:,.2f}
Average Order Value: ${avg_order_value:.2f}
Total Transactions: {total_transactions}
Revenue by Category: {category_revenue.to_dict()}
Revenue % by Category: {category_pct.to_dict()}
Revenue by Season: {season_revenue.to_dict()}
Revenue by Gender: {gender_revenue.to_dict()}
Avg Order Value by Subscription Status: {subscription_avg.to_dict()}
Anomaly Threshold: ${threshold:.2f}
Number of Anomalies: {len(anomalies)}
"""

prompt = f"""You are a retail data analyst. Based on the statistics below, write a business insight report.

STATISTICS:
{stats_summary}

Respond ONLY with valid JSON in exactly this structure, no extra text, no markdown formatting:
{{
  "executive_summary": "3-4 sentence overview",
  "key_findings": ["finding 1", "finding 2", "finding 3"],
  "anomalies_note": "1-2 sentences about the flagged anomalies",
  "recommendation": "1-2 sentence actionable recommendation"
}}
"""

response = client.models.generate_content(
    model="gemini-3.6-flash",
    contents=prompt
)

# Clean and parse the JSON response
raw_text = response.text.strip().replace('```json', '').replace('```', '').strip()
report = json.loads(raw_text)

print("EXECUTIVE SUMMARY:\n", report['executive_summary'], "\n")
print("KEY FINDINGS:")
for f in report['key_findings']:
    print("-", f)
print("\nANOMALIES:\n", report['anomalies_note'], "\n")
print("RECOMMENDATION:\n", report['recommendation'])
````
